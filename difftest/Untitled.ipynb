```python
from greenlamp.differential.discrete import diff
import matplotlib.pyplot as p

t = [0, 0.1, 0.3, 0.4, 0.55, 0.67, 0.71]
x = [23.1, 22.5, 23.5, 21.88, 22.5, 23.5, 24.88]

v = diff(t,x)

fig, plts = p.subplots(2, 1, figsize=(10,6))
plts[0].plot(t,x)
plts[0].set_title("Signal X vs time")
plts[0].set_xlabel("Time (t)")
plts[0].set_ylabel("Signal position (x)")

plts[1].plot(t,v)
plts[1].set_title("Discrete derivative vs time")
plts[1].set_xlabel("Time (t)")
plts[1].set_ylabel("Signal velocity (v)")

fig.tight_layout()

p.show()
```
